### Deep Learning & MNIST

Below, we create two interactive systems:
1. **MNIST Dataset Sample Explorer**: Filter images by choosing target label "Y" and finding the "Xth" occurrence of that digit.
2. **Interactive Prediction Interface**: Interactively select sample test indices to evaluate live predictions against our trained model.

```python
import matplotlib.pyplot as plt
import numpy as np
import ipywidgets as widgets
from IPython.display import display, clear_output
from tensorflow import keras

# Ensure MNIST data is loaded and available for exploring
(mnist_x_train, mnist_y_train), (mnist_x_test, mnist_y_test) = keras.datasets.mnist.load_data()

# Map indexes for each MNIST digit (0-9) to allow finding the X-th sample of Label Y
mnist_digit_indices = {i: np.where(mnist_y_train == i)[0] for i in range(10)}

label_select = widgets.Dropdown(
    options=list(range(10)),
    value=4,
    description='Label Y:',
)

sample_slider = widgets.IntSlider(
    value=2,
    min=1,
    max=100,
    step=1,
    description='Sample X:',
)

image_output = widgets.Output()

def update_mnist_viewer(*args):
    with image_output:
        clear_output(wait=True)
        lbl = label_select.value
        idx_rank = sample_slider.value - 1 # 1-based index to 0-based array index

        available_indices = mnist_digit_indices[lbl]
        if idx_rank >= len(available_indices):
            print(f"Only {len(available_indices)} samples available for label {lbl}.")
            return

        target_train_idx = available_indices[idx_rank]
        img = mnist_x_train[target_train_idx]

        plt.figure(figsize=(3,3))
        plt.imshow(img, cmap='gray')
        plt.title(f"MNIST Sample: Label {lbl} (Occurrence #{sample_slider.value})")
        plt.axis('off')
        plt.show()

label_select.observe(update_mnist_viewer, 'value')
sample_slider.observe(update_mnist_viewer, 'value')

print("Choose Label Y and Occurrence X below to explore:")
display(widgets.VBox([widgets.HBox([label_select, sample_slider]), image_output]))
# Initialize display
update_mnist_viewer()
```

```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers

# Construct and train the Deep Learning model for MNIST so the variable 'model' is defined
model = keras.Sequential([
    layers.Flatten(input_shape=(28, 28)),
    layers.Dense(128, activation='relu'),
    layers.Dense(10, activation='softmax')
])

model.compile(optimizer='adam',
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])

print("Training the Deep Learning model on MNIST dataset...")
model.fit(mnist_x_train, mnist_y_train, epochs=5, batch_size=64)
test_loss, test_acc = model.evaluate(mnist_x_test, mnist_y_test, verbose=0)
print(f"Model trained successfully! Test Accuracy: {test_acc * 100:.2f}%")
```

```python
# 2. Post-Training Interactive Prediction System

predict_output = widgets.Output()
test_index_input = widgets.BoundedIntText(
    value=0,
    min=0,
    max=len(mnist_x_test)-1,
    step=1,
    description='Test Index:',
    disabled=False
)

predict_btn = widgets.Button(
    description='Predict Class',
    button_style='success',
    tooltip='Predict using trained Neural Network'
)

def on_predict_clicked(b):
    with predict_output:
        clear_output(wait=True)
        idx = test_index_input.value
        img = mnist_x_test[idx]

        # Prepare input for trained neural network model
        input_img = np.expand_dims(img, axis=0)
        raw_preds = model.predict(input_img, verbose=0)
        pred_class = np.argmax(raw_preds[0])
        confidence = raw_preds[0][pred_class] * 100

        # Visualizing predicting results
        fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(8, 3))

        ax1.imshow(img, cmap='gray')
        ax1.set_title(f"True Label: {mnist_y_test[idx]}")
        ax1.axis('off')

        y_pos = np.arange(10)
        ax2.barh(y_pos, raw_preds[0], align='center', color='skyblue')
        ax2.set_yticks(y_pos)
        ax2.set_yticklabels(list(range(10)))
        ax2.invert_yaxis()  # top-down label order
        ax2.set_xlabel('Probability')
        ax2.set_title(f"Prediction: {pred_class} ({confidence:.1f}%)")

        plt.tight_layout()
        plt.show()

predict_btn.on_click(on_predict_clicked)

print("Choose any index from the test set (0 to 9999) and hit 'Predict'!")
display(widgets.VBox([widgets.HBox([test_index_input, predict_btn]), predict_output]))
```
