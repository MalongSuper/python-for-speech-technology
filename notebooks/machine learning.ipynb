- Compute Precision, Recall, and F1 by hand for a small confusion

```python
from sklearn.datasets import load_iris

from sklearn.model_selection import train_test_split
X, y = load_iris(return_X_y=True)
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, confusion_matrix

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

# Display the shape
print("X_train:", X_train.shape)
print("X_test:", X_test.shape)
print("y_train:", y_train.shape)
print("y_test:", y_test.shape)

model = RandomForestClassifier()
model.fit(X_train, y_train)
preds = model.predict(X_test)
accuracy = accuracy_score(y_test, preds)
print("Accuracy: %.2f%%" % (accuracy * 100.0))

conf_matrix = confusion_matrix(y_test, preds)
print("Confusion Matrix:\n", conf_matrix)
```

```python
# Calculate the metrics by hand applying the Confusion Matrix
# Here, we use iteration to loop through the unique labels,
# and loop again to compute their values
unique_labels = set(y_test)
for label in unique_labels:
  print("Label:", label)
  tp, tn, fp, fn = 0, 0, 0, 0
  for i in range(len(y_test)):
    # This filtered out entries belong to that specific label
    if y_test[i] == label:
      # This catch the correct predidction
      if preds[i] == label:
        tp += 1
      else:
        fn += 1
    else:  # If not, the label was predicted as something else
      if preds[i] == label:
        fp += 1
      else:
        tn += 1
  precision = tp / (tp + fp)
  recall = tp / (tp + fn)
  f1 = 2 * (precision * recall) / (precision + recall)
  print("Precision:", precision)
  print("Recall:", recall)
  print("F1:", f1)
```
